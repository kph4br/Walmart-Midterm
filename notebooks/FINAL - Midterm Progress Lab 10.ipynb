```python
# TODO
import requests, time

def get_weather(url, params, max_tries=3):
  for attempt in range(1, max_tries+1):
    try:
      response = requests.get(url, params=params)
      response.raise_for_status()
      print("API call worked on attempt", attempt)
      return response.json()
    except Exception as e:
      print("Attempt", attempt, "failed:", e)
      if attempt < max_tries:
        time.sleep(2 * attempt)
  return None

stores = pd.read_csv(f'{REPO_PATH}/data/store_locations.csv')
stores['store_id'] = stores['store_id'].apply(clean_store_id)
stores = stores.drop_duplicates(subset='store_id')
stores.loc[stores['store_id'] == 'FL-619', ['latitude', 'longitude']] = [30.4383, -84.2807]

# daily weather for each store
url = "https://archive-api.open-meteo.com/v1/archive"
all_weather = []
for i in range(len(stores)):
  store = stores.iloc[i]
  params = {
    "latitude": store['latitude'],
    "longitude": store['longitude'],
    "start_date": "2024-09-01",
    "end_date": "2024-09-30",
    "daily": "precipitation_sum,wind_speed_10m_max",
    "hourly": "pressure_msl",
    "timezone": "America/New_York"
  }
  data = get_weather(url, params)
  if data is not None:
    w = pd.DataFrame({
        'date': pd.to_datetime(data['daily']['time']),
        'precip_mm': data['daily']['precipitation_sum'],
        'max_wind_kmh': data['daily']['wind_speed_10m_max'],
    })
    hourly = pd.DataFrame({'time': pd.to_datetime(data['hourly']['time']),
                           'pressure': data['hourly']['pressure_msl']})
    hourly['date'] = hourly['time'].dt.normalize()
    w['min_pressure_hpa'] = w['date'].map(hourly.groupby('date')['pressure'].min())
    w['store_id'] = store['store_id']
    w['source'] = 'live: Open-Meteo archive API'
    all_weather.append(w)
  else:
    print("Skipping", store['store_id'], "because the call failed")
  time.sleep(1)

weather = pd.concat(all_weather, ignore_index=True)
print('weather rows:', len(weather), '| stores:', weather['store_id'].nunique(), '| source:', weather['source'].iloc[0])

# validated join: one row per store per day
store_day = d.groupby(['store_id', 'date'])['quantity'].sum().reset_index(name='units')
rows_before = len(store_day)
joined = store_day.merge(weather, on=['store_id', 'date'], how='left', validate='one_to_one')
rows_after = len(joined)
print('rows before join:', rows_before, '| after join:', rows_after,
      '| store-days missing weather:', joined['precip_mm'].isna().sum())
assert rows_before == rows_after

# correlation: same-day weather, and weather 2 days later (people buy BEFORE the storm hits)
joined = joined.sort_values(['store_id', 'date'])
joined['precip_2_days_later'] = joined.groupby('store_id')['precip_mm'].shift(-2)
joined['wind_2_days_later'] = joined.groupby('store_id')['max_wind_kmh'].shift(-2)
print('\ndays:', joined['date'].nunique(), '| stores:', joined['store_id'].nunique())
for col in ['precip_mm', 'max_wind_kmh', 'precip_2_days_later', 'wind_2_days_later']:
    sub = joined.dropna(subset=[col])
    print(f'units vs {col}: r = {sub["units"].corr(sub[col]):.2f}  (n = {len(sub)} store-days)')

# chart: all stores combined, units vs average rain
state = joined.groupby('date').agg(units=('units', 'sum'), precip_mm=('precip_mm', 'mean'))
fig, ax1 = plt.subplots(figsize=(9, 3.8))
ax1.bar(state.index, state['precip_mm'], color='#9db4cf', label='avg rain across stores (mm)')
ax1.set_ylabel('avg rain (mm)')
ax2 = ax1.twinx()
ax2.plot(state.index, state['units'], color='#b4553f', marker='o', ms=3, label='units sold (all stores)')
ax2.set_ylabel('units sold')
ax1.axvline(pd.Timestamp('2024-09-26'), ls='--', color='gray', lw=1)
ax1.set_title('Sales Peak Before Rain', loc='left', fontsize=11)
fig.legend(frameon=False, fontsize=9, loc='upper left', bbox_to_anchor=(0.07, 0.88))
ax1.tick_params(axis='x', rotation=45)
plt.tight_layout(); plt.show()
```

```python
# Chart for slidedeck
import seaborn as sns

# Average weather across all 12 stores for each day
state_weather = weather.groupby('date')[['precip_mm', 'max_wind_kmh', 'min_pressure_hpa']].mean()
combo = cat_daily.join(state_weather)

# Correlation of each weather measure with each category's sales
corr = combo.corr().loc[cat_daily.columns, state_weather.columns]
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title('Weather vs. sales by category (30 days)')
plt.tight_layout()
plt.show()

# Compare sales today with the weather 0-5 days LATER
total_sales = cat_daily.sum(axis=1)
for days in range(0, 6):
    rain = total_sales.corr(state_weather['precip_mm'].shift(-days))
    wind = total_sales.corr(state_weather['max_wind_kmh'].shift(-days))
    pressure = total_sales.corr(state_weather['min_pressure_hpa'].shift(-days))
    print(f'weather {days} days later -> rain r = {rain:.2f}, wind r = {wind:.2f}, pressure r = {pressure:.2f}')
```

**Q3 answer, with its caveat:** Store sales correlated strongly with the forcast represented two days in the future (r=0.68 for precipitation, r=0.62 for peak wind), rather than same day conditions (r=.12), showing the customer purchasing behavior is anticipatory. The key caveat is that this data is calculated across 12 stores over 30 days (360 store days) and a single event like Hurricane Helene drove the macro preparation rush and the subsequent weather shock.

To verify this was true customer behavior patterns, we audited the distance calculations using raw latitude/longitude coordinates, confirmed the manual fix on store Fl-619 (Tallahassee), and verified that daily sales volumes were properly grouped by store ID before computing the pre-storm multiples.

b) The reported lead-time correlation of r=0.68 between store unit sales and rainfall two days later (r=0.68). While it looks like a strong pattern, all the data comes from one hurricane hitting Florida over a single month. Because we are looking at one event, rather than several separate storms the sales rush and naturally lined up together. We cannot prove this pattern would happen the same way for future storms.
